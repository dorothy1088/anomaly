# 통계 규칙, STL 잔차,// isolationforest(다변량)
이상치 다른 값들과 동떨어진 값 자체, 이상징후는 그 중에서 평소의 패턴으로 설명되지 않기 때문에 원인을 확인해 볼만한 가치가 있는 신호를 의미한다.

통계규칙 이상치를 찾아준다. 그것이 이상징후인지는 사람이 원인과 대조해서 판단한다. 모든 튀는 값이 문제는 아니고 반대로 문제가 있는 날에 무조건 값이 튀는 것도 아니기 때문이다.

부산항 데이터에서 값이 튄다라고 한다면 어떤 원인이 있을지 고민해 볼 수 있다.

- 데이터 오류 : 입항일시 1950년, 7800만톤 화물
- 설명되는 이상 : 명절, 태풍 같은 특정 요인
- 설명 되지 않는 이상(이상징후): 공휴일도 아닌데 평일에 데이터가 급감하거나 급증하거나

이상징후 탐지에는 정답 라벨이 없다. 이 부분이 예측과 가장 다른 부분이다. 이 날이 진짜 이상이었다는 정확도로 계산할 수 없고 탐지 결과를 알려진 사건과 대조해서 설명되는 비율과 설명 안되는 날의 목록으로 평가할 수 밖에 없기 때문이다.

## Z-SCORE
값이 얼마나 튀는지를 측정하는 가장 기본적인 방법은 평균에서 표준편차 몇 개만큼 떨어져 있는지를 확인하는 방법이다. 이 방법이 Z-SCORE표준점수라고 일컫는다.

Z-SCORE는 단점이 있는데 평균과 표준편차가 이상치에 끌려다닌다라는 것이다. 이상치가 기준을 흐려서 스스로를 숨기는 현상을 가림 효과MASKING이라고 일컫는다. 데이터가 적거나 이상치가 여러 개일 때 영향이 크다.

## IQR 규칙
사분위수 범위IQR로 이상치를 찾을 수 있다. Q1-1.5*IQR보다 작거나 Q3+1.5*IQR보다 크거나 한 값을 이상치로 판단하는 방법

사분위수는 값을 크기순으로 늘어놓은 위치로 결정되기 때문에 극단값이 아무리 커도 Q1과 Q3은 거의 움직이지 않는다. 이 성질을 강건하다ROBUST라고 한다.


## STL 잔차 탐지
STL 분해는 seasonal-trend decomposition using loess를 의미한다. 시계열을 추세, 계절, 나머지 나누는 방법, 나머지가 잔차residual가 된다.
